```python
import time

from pytweezer.drivers.imagemX2 import ImagEMX2Camera
```

```python
cam = ImagEMX2Camera()
```

```python
cam.enable_em_gain()
cam.enable_direct_em_gain()
cam.set_sensitivity(1)
cam.set_trigger_source("software")
cam.setup_acquisition("snap", 1)
cam.set_exposure_time(0.001)
cam.set_roi(0, 0, 512, 512)
cam.start_acquisition()
```

```python
t0 = time.time()
n = 100
for i in range(n):
    cam.start_acquisition()
    cam._backend.send_software_trigger()
    img = cam.acquire_single_frame()
    time.sleep(
        0.15
    )  # any less than that and we get lost frames and the camera needs a close() and re-open() to work again. This is a bug in the camera driver, not pytweezer.
print(f"Time for {n} frames: {time.time() - t0:.3f} s")
```

```python
cam.stop_acquisition()
```

```python
cam.close()
```

```python
from pytweezer.servers.device_client import get_device
```

```python
cam: ImagEMX2Camera = get_device("rbhamcam")
```

```python
cam.enable_em_gain()
cam.enable_direct_em_gain()
cam.set_sensitivity(1)
cam.set_trigger_source("software")
cam.setup_acquisition("snap", 1)
cam.set_exposure_time(0.00001)
cam.set_roi(0, 1024, 0, 1024)
cam.start_acquisition()
```

```python
t0 = time.time()
n = 10
for i in range(n):
    cam.start_acquisition()
    cam.trigger()
    img = cam.acquire_single_frame(broadcast=True)
    time.sleep(
        0.15
    )  # any less than that and we get lost frames and the camera needs a close() and re-open() to work again. This is a bug in the camera driver, not pytweezer.
print(f"Time for {n} frames: {time.time() - t0:.3f} s")
```
